## Exercice 3 — Notation polonaise inversée

### Correction — Notes prises en cours

```python
def eval_pol_inv(ch):
    exp = ch.split()
    stack = []

    for e in exp:
        if e == "+":
            stack.append(int(stack.pop()) + int(stack.pop()))

        elif e == "-":
            premier = int(stack.pop())
            second = int(stack.pop())
            stack.append(second - premier)

        elif e == "":
            stack.append(int(stack.pop()) int(stack.pop()))

        else:
            stack.append(e)

    return stack[0]
# on peut utilsier eval
```

```python
class Noeud:
    def __init__(self, elm, suiv):
        self.element = elm
        self.suivant = suiv


class Liste:
    def __init__(self):
        self.tete = None

    # 1.
    def est_vide(self):
        return self.tete is None

    def ajoute(self, elm):
        self.tete = Noeud(elm, self.tete)

    def tete_liste(self):
        return self.tete.element

    def reste(self):
        self.tete = self.tete.suivant
        return self

    # 2.a
    def __trouve_iem__(self, i):
        courant = self.tete

        while i>0 and courant is not None:
            courant = courant.suivant
            i -= 1
            return courant
        #for _ in range(i):
        #    courant = courant.suivant

        #return courant.element

    def __setitem__(self, i, valeur):
        courant = self.tete

        for _ in range(i):
            courant = courant.suivant

        courant.element = valeur

    # 2.b
    def __len__(self):
        courant = self.tete
        taille = 0

        while courant is not None:
            taille += 1
            courant = courant.suivant

        return taille

    # 2.c
    def index(self, elm):
        courant = self.tete
        i = 0

        while courant is not None:
            if courant.element == elm:
                return i

            courant = courant.suivant
            i += 1

        return -1

    
```
